```python
import numpy as np
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples=100, n_informative=1, n_features=2, n_redundant=0, n_clusters_per_class=1, n_classes=2, hypercube=False, class_sep=15, random_state=41)
```

```python
print(X.shape)
print(y.shape)
```

```python
print(X[:5])
print(y[:5])
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,6))
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter')
plt.show()
```

```python
def perceptron(X,y):
  w1=w2=b=1
  lr=0.1

  for i in range(1000):
    for i in range(X.shape[0]):
      z = w1*X[i][0] + w2*X[i][1] + b

      if z*y[i] < 0:
        w1 += lr*y[i]*X[i][0]
        w2 += lr*y[i]*X[i][1]
        b += lr*y[i]

  return w1,w2,b
```

```python
w1,w2,b = perceptron(X,y)
```

```python
print(w1)
print(w2)
print(b)
```

```python
m = - (w1/w2)
c = - (b/w2)

print(m)
print(c)
```

```python
Xinput = np.linspace(-3,3,100)
yinput = m*Xinput + c

plt.figure(figsize=(8,6))
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter')
plt.plot(Xinput, yinput, color='red')
plt.show()
```
