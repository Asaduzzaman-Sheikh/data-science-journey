```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
or_data = pd.DataFrame()
and_data = pd.DataFrame()
xor_data = pd.DataFrame()
```

```python
or_data['A'] = [0, 0, 1, 1]
or_data['B'] = [0, 1, 0, 1]
or_data['Y'] = [0, 1, 1, 1]
```

```python
and_data['A'] = [0, 0, 1, 1]
and_data['B'] = [0, 1, 0, 1]
and_data['Y'] = [0, 0, 0, 1]
```

```python
xor_data['A'] = [0,0,1,1]
xor_data['B'] = [0,1,0,1]
xor_data['Y'] = [0,1,1,0]
```

```python
or_data
```

```python
and_data
```

```python
xor_data
```

```python
fig, ax = plt.subplots(1,3, figsize=(15,5))

sns.scatterplot(or_data, x='A', y='B', hue='Y', ax=ax[0])
ax[0].set_title('OR Gate')

sns.scatterplot(and_data, x='A', y='B', hue='Y', ax=ax[1])
ax[1].set_title('AND Gate')

sns.scatterplot(xor_data, x='A', y='B', hue='Y', ax=ax[2])
ax[2].set_title('XOR Gate')

plt.tight_layout()
plt.show()
```

```python
from sklearn.linear_model import Perceptron


perceptron1 = Perceptron()
perceptron2 = Perceptron()
perceptron3 = Perceptron()


perceptron1.fit(or_data.iloc[:,:2].values, or_data.iloc[:,-1].values)
perceptron2.fit(and_data.iloc[:,:2].values, and_data.iloc[:,-1].values)
perceptron3.fit(xor_data.iloc[:,:2].values, xor_data.iloc[:,-1].values)
```

```python
print(f"Perceptron1 coeffiecinet: {perceptron1.coef_}")
print(f"Perceptron1 intercept: {perceptron1.intercept_}")
print('------------------------------------------------------------')
print(f"Perceptron2 coeffiecinet: {perceptron2.coef_}")
print(f"Perceptron2 intercept: {perceptron2.intercept_}")
print('------------------------------------------------------------')
print(f"Perceptron3 coeffiecinet: {perceptron3.coef_}")
print(f"Perceptron3 intercept: {perceptron3.intercept_}")
print('------------------------------------------------------------')
```

```python
x=np.linspace(-1,1,5)
y=-x+1

x1=np.linspace(-1,1,5)
y1=-x+0.5
```

```python
fig, ax = plt.subplots(1,3, figsize=(15,5))

sns.scatterplot(or_data, x='A', y='B', hue='Y', ax=ax[0])
ax[0].plot(x1,y1)
ax[0].set_title('OR Gate')

sns.scatterplot(and_data, x='A', y='B', hue='Y', ax=ax[1])
ax[1].plot(x,y)
ax[1].set_title('AND Gate')

sns.scatterplot(xor_data, x='A', y='B', hue='Y', ax=ax[2])
ax[2].plot(x,y)
ax[2].set_title('XOR Gate')

plt.tight_layout()
plt.show()
```
