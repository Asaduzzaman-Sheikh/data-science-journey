# Handling Mixed Variable
**Handling mixed variables in machine learning** requires strategic feature engineering or specialized algorithms because models cannot inherently process combinations of letters, symbols, and numbers within the same input space. Mixed variables usually present themselves in two main ways: **alphanumeric strings combined in a single cell (e.g., ticket codes like "A105") or columns containing entirely numeric values in some rows and categorical values in others.**  

---
## 🛠️ Feature Engineering Strategies
The most common approach is to split the mixed variable into distinct numerical and categorical columns before feeding them into your modeling pipeline  

1. **Splitting Alphanumeric Strings (Regex Extraction)**: When text and numbers are merged together, use regular expressions (regex) to extract the latent components into separate, clean features
   * Text/Category extraction: Extract letters to identify categories like "Deck Level" or "Region".
   * Numerical extraction: Extract digits to represent continuous values like "Room Number" or "ID sequence"
2.  **Segregating Conditional Rows**: If a column contains pure numbers in some rows and string categories in others, use conditional masks to separate them, filling the alternate entries with structural missing placeholders (NaN or 'Unknown')
---

## 🧪 Downstream Encoding & Preprocessing
Once the mixed variable is split into standalone columns, you must apply standard preprocessing techniques to ensure compatibility with your model:
* For the Numerical Component: Handle missing values with median/mean imputation, then scale the values using techniques like `StandardScaler` or `MinMaxScaler`
* For the Categorical Component: Impute missing categories with a string placeholder like `'Missing'` or `'Unknown'`. Transform them using **One-Hot Encoding** for low-cardinality features or **Target Encoding** if the column has high cardinality
---

### Spliting Alphanumerical Strings (Regex Extraction)

```python
import pandas as pd

df = pd.DataFrame({'Cabin': ['C85', 'E36', 'A10', 'C123']})

# Extract categorical letters and numerical digits into separate features
df['Cabin_letter'] = df['Cabin'].str.extract(r'([A-Za-z]+)')
df['Cabin_number'] = df['Cabin'].str.extract(r'(\d+)').astype(float)
```

### Segregating Conditional Rows

```python
import numpy as np

df = pd.DataFrame({'Ticket': ['113803', 'A/5 21171', 'STON/O2. 3101282', '248706']})

# Isolate numeric values; non-numeric entries become NaN
df['Ticket_numeric'] = pd.to_numeric(df['Ticket'], errors='coerce')

# Isolate string values; pure numeric entries become NaN or a placeholder
df['Ticket_categorical'] = np.where(df['Ticket_numeric'].isnull(), df['Ticket'], np.nan)
```
