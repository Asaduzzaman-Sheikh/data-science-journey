# Handling Date and Time Variable
Handling date and time variables involves `converting raw text into proper timestamp formats and extracting useful parts like day, month, year, or time elapsed`

---
**Main Steps to Handle Date and Time** 
* `Convert Data Type:` Change string or object columns into formal datetime objects using tools like Pandas to_datetime so you can run calculations.
* `Extract Features:` Pull hidden details from a single date column, such as the hour, day of the week, month, year, or whether a day is a weekend.
* `Calculate Differences:` Subtract an start time from an end time to find the duration or time delta between two events.
* `Manage Time Zones:` Set or convert all timestamps to a single unified time zone to avoid errors in your data.

```python
import numpy as np
import pandas as pd
```

```python
date = pd.read_csv('orders.csv')
time = pd.read_csv('messages.csv')
```

```python
date.head()
```

```python
time.head()
```

## Working with Date

```python
date.info()
```

```python
date['date'] = pd.to_datetime(date['date'])
date.info()
```

```python
# Extract year
date['year'] = date['date'].dt.year
date.head()
```

```python
# Extract month
date['month'] = date['date'].dt.month
date.sample(5)
```

```python
# Extract month name
date['month_name'] = date['date'].dt.month_name()
date.sample(5)
```

```python
# Extract days
date['days'] = date['date'].dt.day
date.sample(5)
```

```python
# Extract days name
date['day_name'] = date['date'].dt.day_name()
date.sample(5)
```

```python
# Day of the week
date['day_of_the_week'] = date['date'].dt.dayofweek
date.sample(5)
```

```python
date['is_weekend'] = np.where(date['day_name'].isin(['Sunday','Saturday']), 1, 0)
date.sample(5)
```

```python
# Extract week of the year
date['week_of_the_year'] = date['date'].dt.isocalendar().week
date.sample(5)
```

```python
# Extract quarter
date['quarter'] = date['date'].dt.quarter
date.sample(5)
```

```python
# Extract semester
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)
date.sample(5)
```

```python
# Extract time elasped between dates
import datetime

today = datetime.datetime.today()
today
```

```python
today - date['date']
```

```python
(today - date['date']).dt.days
```

```python
# Months passed
np.round((today - date['date']) / np.timedelta64(1,'D')/ 30.4375,1)
```

## Working with Time

```python
time.info()
```

```python
time['date'] = pd.to_datetime(time['date'])
```

```python
time.info()
```

```python
time['hours'] = time['date'].dt.hour
time['minutes'] = time['date'].dt.minute
time['seconds'] = time['date'].dt.second
time.sample(5)
```

```python
# Extract time part form the date
time['time'] = time['date'].dt.time
time.head()
```

```python
# Time difference
today - time['date']
```

```python
(today - time['date']) / np.timedelta64(1, 'h')
```

```python
(today - time['date']) / np.timedelta64(1, 'm')
```

```python
(today - time['date']) / np.timedelta64(1, 's')
```
