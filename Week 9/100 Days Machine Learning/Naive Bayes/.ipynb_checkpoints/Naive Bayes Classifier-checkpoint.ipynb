# Naive Bayes Classifier
## 💡 Core Idea
Choose the class that most likely generated the observed features.
## 📐 Bayes’ Theorem for Classification
$$P(C \mid X) \propto P(C) \prod_{i} P(x_i \mid C)$$ 
## 🔑 Key Terms


* Prior $P(C)$: Belief before seeing features.
* Likelihood $P(x_i \mid C)$: How likely a feature is for a given class.
* Posterior $P(C \mid X)$: Updated belief after observing features.
* Naive Assumption: Features are assumed to be independent given the class. 


## ⚖️ Prediction Rule
$$C^* = \arg\max_{C} P(C) \prod_{i} P(x_i \mid C)$$ 
Log-Space Transformation (Used to avoid numerical underflow): 
$$\log P(C) + \sum_{i} \log P(x_i \mid C)$$ 
------------------------------
## 📦 Common Variants


* Multinomial: Based on word counts (frequently used in text classification).
* Bernoulli: Based on word presence or absence (binary indicators).
* Gaussian: Designed for continuous features. 


------------------------------
## ⚖️ Pros & Cons

| Pros | Cons |
|---|---|
| ⚡ Fast and simple | ❌ Independence assumption is often false |
| 📊 Works well with small data | 🎯 Probabilities can be overconfident |
| 🌌 Great for high-dimensional text | 🧼 Needs smoothing (e.g., Laplace) for zero counts |
| 🏷️ Handles multi-class tasks easily | |

------------------------------
## 🛠️ Use Cases


* Spam filtering
* Sentiment analysis
* Document classification
* Medical triage 

🧠 Memory Hook: Prior belief + independent clue votes = posterior decision.

```python
df = pd.read_csv('play_tennis.csv')
```

```python
df = df.drop(columns = 'day')
df.head()
```

```python
# Problem Statement: Given that outlook = sunny, temp = hot, humidity = high, wind = weak
# So, what is the probability to paly or not play

"""
    Solution:
    P(Yes|outlook = sunny, temp = hot, humidity = high, wind = weak) = P(outlook = sunny| Yes)*P(temp = hot|Yes)*P(humidity = high|Yes)*P(wind = weak|Yes)*P(Yes)
    P(No|outlook = sunny, temp = hot, humidity = high, wind = weak) = P(outlook = sunny|No)*P(temp = hot|No)*P(humidity = high|No)*P(wind = weak|No)*P(No)

    Now, compare which probability is high -> result
    

"""

# In training phase, Naive Bayes -> create a lookup table (dictionary)
# In testing phase, Naive Baye -> search in the lookup table to find the specific case's probability
print()
```

```python
# For outlook column
pd.crosstab(df['outlook'], df['play'])
```

```python
p_Overcast_given_yes = 4/9
p_Rain_given_yes = 3/9
p_Sunnuy_given_yes = 2/9


p_Overcast_given_no = 0
p_Rain_given_no = 2/5
p_Sunny_given_no = 3/5
```

```python
# For Temperature column
pd.crosstab(df['temp'], df['play'])
```

```python
p_Cool_given_yes = 3/9
p_Hot_given_yes = 2/9
p_Mild_given_yes = 4/9


p_Cool_given_no = 1/5
p_Hot_given_no = 2/5
p_Mild_given_no = 2/5
```

```python
# For Humidity column
pd.crosstab(df['humidity'], df['play'])
```

```python
p_High_given_yes = 3/9
p_Normal_given_yes = 6/9


p_High_given_no = 4/5
p_Normal_given_no = 1/5
```

```python
# For Wind column
pd.crosstab(df['wind'], df['play'])
```

```python
p_Strong_given_yes = 3/9
p_Weak_given_yes = 6/9


p_Strong_given_no = 3/5
p_Weak_given_no = 2/5
```

```python
# For the target column
df['play'].value_counts()
```

```python
p_yes = 9/14
p_no = 5/14
```

```python
# Given that outlook = sunny, temp = hot, humidity = high, wind = weak
P_YES = p_Sunnuy_given_yes*p_Hot_given_yes*p_High_given_yes*p_Weak_given_yes*p_yes
P_NO = p_Sunny_given_no*p_Hot_given_no*p_High_given_no*p_Weak_given_no*p_no
```

```python
print(f"Probability to play(YES): {P_YES}")
print(f"Probability not to play(NO): {P_NO}")
```
