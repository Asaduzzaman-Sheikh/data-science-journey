## 1. Introduction

### What is Optuna?
Optuna is an **open-source hyperparameter optimization framework** designed for machine learning and general optimization tasks. It automates the search for the best hyperparameters by:

- **Learning from past trials** (Bayesian optimization via TPE)
- **Pruning unpromising trials early** to save compute
- **Providing rich visualization** for analysis
- **Scaling horizontally** across distributed workers

### Why Optuna?
| Feature | Benefit |
| :--- | :--- |
| Define-by-run API | Search space is defined dynamically in Python code |
| State-of-the-art samplers | TPE, CMA-ES, NSGA-II, GPSampler |
| Built-in pruning | Median, Hyperband, SuccessiveHalving |
| Framework agnostic | Works with sklearn, PyTorch, TensorFlow, XGBoost, LightGBM, etc. |
| Distributed | Scale across many machines via shared storage |
| Visualization | Rich plots out-of-the-box + Optuna Dashboard |

## 3. Core Concepts

### 3.1 The Two Pillars

| Concept | Description | Created With |
| :--- | :--- | :--- |
| **Study** | A complete optimization project. Manages trials, stores history, tracks the best result. | `optuna.create_study()` |
| **Trial** | A single evaluation of the objective with one set of hyperparameters. | Auto-created by `study.optimize()` |

### 3.2 The Objective Function
- A Python callable that takes a `trial` object and returns a scalar (or tuple for multi-objective).
- **Contains the search space** via `trial.suggest_*()` calls.
- **Returns the value** Optuna should minimize/maximize.

### 3.3 Optimization Direction
```python
study = optuna.create_study(direction="minimize")  # or "maximize"
```

## 4. Your First Optimization

```python
import optuna

# 1. Define the objective
def objective(trial):
    x = trial.suggest_float("x", -10, 10)
    return (x - 2) ** 2

# 2. Create a study
study = optuna.create_study(direction="minimize")

# 3. Optimize
study.optimize(objective, n_trials=100)

# 4. Inspect results
print(f"Best params: {study.best_params}")
print(f"Best value:  {study.best_value}")
print(f"Best trial:  {study.best_trial.number}")
```

### What Happens Under the Hood
1. Study is created with a default sampler (`TPESampler`) and default pruner (`MedianPruner`-like behavior via `NopPruner` unless set).
2. `optimize()` runs `objective` 100 times.
3. In each trial, `trial.suggest_float` proposes a value.
4. The study records the result and updates its internal model.
5. Best parameters are stored in `study.best_params`.

## 5. Defining the Search Space

Optuna uses a **define-by-run** paradigm — the search space emerges from code executed per trial.

### 5.1 The `suggest_*` API

| Method | Use Case | Key Arguments | Example |
| :--- | :--- | :--- | :--- |
| `suggest_float(name, low, high)` | Continuous values | `log`, `step` | `trial.suggest_float("lr", 1e-5, 1e-1, log=True)` |
| `suggest_int(name, low, high)` | Discrete integers | `step`, `log` | `trial.suggest_int("n_layers", 1, 5)` |
| `suggest_categorical(name, choices)` | Enumerated options | `choices` | `trial.suggest_categorical("opt", ["adam", "sgd"])` |

### 5.2 Log-Scale Sampling (Critical!)

For parameters spanning **multiple orders of magnitude** (learning rate, regularization strength, gamma in SVM, etc.):

```python
learning_rate = trial.suggest_float("lr", 1e-5, 1e-1, log=True)
```

✅ Samples uniformly in *log space* → small values get fair representation.

❌ Without `log=True`, values near `high` dominate — bad for exponential parameters.

### 5.3 Conditional / Dependent Search Spaces

Since the search space is just Python, you can branch:

```python
def objective(trial):
    optimizer = trial.suggest_categorical("optimizer", ["adam", "sgd"])
    lr = trial.suggest_float("lr", 1e-5, 1e-1, log=True)
    
    if optimizer == "adam":
        beta1 = trial.suggest_float("beta1", 0.8, 0.999)
        beta2 = trial.suggest_float("beta2", 0.9, 0.9999)
    else:  # sgd
        momentum = trial.suggest_float("momentum", 0.0, 0.99)
    
    return train_and_evaluate(optimizer, lr, ...)
```

### 5.4 Fixed Parameters & Tips

- Use `trial.suggest_categorical("flag", [True, False])` for booleans.
- For a parameter you don't want to tune but want logged: `trial.set_user_attr("key", value)`.
- **Don't change a parameter's name or type mid-study** — it breaks the sampler's model.

## 6. Samplers

Samplers decide **which hyperparameters to try next** based on past results.

### 6.1 Setting a Sampler

```python
study = optuna.create_study(
    direction="minimize",
    sampler=optuna.samplers.TPESampler(seed=42, n_startup_trials=10)
)
```

### 6.2 Sampler Comparison

| Sampler | Best For | Avoid When |
| :--- | :--- | :--- |
| **`TPESampler`** (default) | General purpose, mixed param types | < 20 trials |
| **`RandomSampler`** | Baseline, embarrassingly parallel setups | Almost always |
| **`CmaEsSampler`** | Many continuous params (neural net HPO) | Categorical dominates |
| **`GridSampler`** | Exhaustive small grids | Combinatorial explosion |
| **`NSGAIISampler`** | Multi-objective (2–3 objectives) | Single objective |
| **`NSGAIIISampler`** | Multi-objective (many objectives) | Single objective |
| **`GPSampler`** | Expensive, few-trial budgets (< 15 params) | High-dim spaces |
| **`QMCSampler`** | Quasi-Monte Carlo for good coverage | Complex interactions |

### 6.3 TPESampler Key Arguments

```python
optuna.samplers.TPESampler(
    seed=42,                   # Reproducibility
    n_startup_trials=10,       # Random exploration before TPE kicks in
    n_ei_candidates=24,        # Number of candidates for EI maximization
    multivariate=True,         # Model parameter dependencies
    group=False,               # Enable grouped TPE
    consider_prior=True,       # Use prior for cold-start
    prior_weight=1.0
)
```

### 6.4 CMA-ES (for continuous spaces)

```python
study = optuna.create_study(
    sampler=optuna.samplers.CmaEsSampler(seed=42)
)
```
> Requires `cmaes` package: `pip install cmaes`

## 7. Pruners

Pruners **stop unpromising trials early** — critical for expensive models.

### 7.1 How Pruning Works

1. Create a study with a pruner.
2. Inside the objective, **report intermediate values** at each step/epoch.
3. **Check `trial.should_prune()`**; if `True`, raise `optuna.TrialPruned()`.

```python
def objective(trial):
    lr = trial.suggest_float("lr", 1e-5, 1e-1, log=True)
    model = build_model(lr)
    
    for epoch in range(n_epochs):
        train_one_epoch(model)
        val_loss = evaluate(model)
        
        trial.report(val_loss, epoch)   # Report step value
        
        if trial.should_prune():        # Ask pruner
            raise optuna.TrialPruned()  # Stop trial
    
    return val_loss
```

### 7.2 Pruner Comparison

| Pruner | Strategy | Use When |
| :--- | :--- | :--- |
| **`MedianPruner`** | Prune if worse than median of past trials at same step | Safe default |
| **`HyperbandPruner`** | Multi-bracket successive halving (ASHA) | Large budget, want aggressive pruning |
| **`SuccessiveHalvingPruner`** | Iteratively keep top 1/η of trials | Similar to Hyperband, single bracket |
| **`PercentilePruner`** | Prune if worse than a percentile | Tunable strictness |
| **`ThresholdPruner`** | Prune if value crosses a threshold | Hard cutoff known |
| **`PatientPruner`** | Prune if no improvement after N steps | Noisy metrics |
| **`NopPruner`** | Never prune | Debugging / small studies |

### 7.3 Example: Hyperband

```python
study = optuna.create_study(
    direction="minimize",
    pruner=optuna.pruners.HyperbandPruner(
        min_resource=1,      # Min epochs
        max_resource=100,    # Max epochs
        reduction_factor=3
    )
)
```

### 7.4 MedianPruner Arguments

```python
optuna.pruners.MedianPruner(
    n_startup_trials=5,   # Don't prune first N trials
    n_warmup_steps=10,    # Don't prune before step N
    interval_steps=1,     # Check every N steps
    n_min_trials=1        # Min trials at step before pruning
)
```

## 9. Advanced Features

### 9.1 Multi-Objective Optimization

Find the **Pareto front** of competing objectives (e.g., accuracy ↑ vs. latency ↓).

```python
study = optuna.create_study(
    directions=["maximize", "minimize"],  # [accuracy, latency]
    sampler=optuna.samplers.NSGAIISampler()
)

def objective(trial):
    lr = trial.suggest_float("lr", 1e-5, 1e-1, log=True)
    model = train(lr)
    return accuracy(model), latency(model)

study.optimize(objective, n_trials=200)

# Access Pareto-optimal trials
for t in study.best_trials:
    print(t.number, t.values, t.params)
```

### 9.2 Distributed / Parallel Optimization

Use a **shared storage** so multiple workers pull trials from the same study.

```python
# Every worker runs this same code
study = optuna.create_study(
    study_name="distributed_hpo",
    storage="sqlite:///optuna_study.db",   # or "postgresql://..."
    load_if_exists=True,
    direction="minimize"
)
study.optimize(objective, n_trials=50)  # Each worker does 50 trials
```

**Storage Backends:**
| Backend | Use Case |
| :--- | :--- |
| SQLite | Local, single machine, multiple processes |
| PostgreSQL / MySQL | Production, multi-machine |
| Redis | Fast, in-memory (via `optuna[redis]`) |

> ⚠️ SQLite has limited concurrency — for many workers use PostgreSQL.

### 9.3 The Ask-and-Tell Interface (Low-Level API)

Full control: separate parameter suggestion from evaluation.

```python
study = optuna.create_study()

for _ in range(100):
    trial = study.ask()                                    # Ask for params
    x = trial.suggest_float("x", -10, 10)
    y = external_function(x)                               # Evaluate externally
    study.tell(trial, y)                                   # Report result
```

**Use cases:**
- Integrating with external job schedulers (SLURM, Kubernetes)
- Batch evaluation
- Asynchronous workflows

### 9.4 Warmstarting with Known-Good Parameters

Pre-seed the study with promising configurations:

```python
study.enqueue_trial({"lr": 0.01, "batch_size": 64})
study.enqueue_trial({"lr": 0.001, "batch_size": 128})
study.optimize(objective, n_trials=100)
```

> Great when you have prior knowledge or hyperparameters from earlier experiments.

### 9.5 Resuming a Study

```python
study = optuna.load_study(
    study_name="my_study",
    storage="sqlite:///optuna_study.db"
)
study.optimize(objective, n_trials=50)  # Continue optimizing
```

### 9.6 User Attributes (Metadata)

```python
# Inside objective

# Later
```

### 9.7 Callbacks

Run code after each trial (e.g., early stopping, logging, notifications):

```python
def early_stop_callback(study, trial):
    if study.best_value < 0.01:
        study.stop()

study.optimize(objective, n_trials=1000, callbacks=[early_stop_callback])
```

### 9.8 Trial Logging (Silence Output)

```python

    objective,

    show_progress_bar=True,
    callbacks=[lambda s, t: None]  # Suppress default logs
)
```

## 10. Best Practices

### 10.1 Search Space Design
- ✅ **Use `log=True`** for exponential parameters (lr, gamma, regularization).
- ✅ **Set sensible bounds** — sampling range [0, 1e10] wastes trials.
- ✅ **Use conditional spaces** to avoid meaningless combinations.
- ❌ **Don't tune parameters you know are unimportant** — wastes budget.
- ❌ **Don't change parameter names/types mid-study** — breaks the sampler.

### 10.2 Trial Budget
- Start with **50–100 trials** to get a sense of convergence.
- Use `plot_optimization_history` to check if more trials will help.
- Set a **`timeout`** for safety: `study.optimize(objective, timeout=3600)`

### 10.3 Pruning
- **Always** use pruning for expensive training loops (neural nets, gradient boosting).
- Report intermediate values at meaningful intervals (each epoch, every N steps).
- Use `HyperbandPruner` or `SuccessiveHalvingPruner` for aggressive savings.

### 10.4 Reproducibility
```python
sampler = optuna.samplers.TPESampler(seed=42)
study = optuna.create_study(sampler=sampler)
```
Plus set seeds in NumPy, PyTorch, TensorFlow, and use deterministic data splits.

### 10.5 Persistence
- **Always** pass a `storage` URL — never rely on in-memory studies for anything serious.
- SQLite for local, PostgreSQL for teams/production.

### 10.6 Parallelization
- Use **TPE with `n_startup_trials`** set properly (~10–20% of total trials).
- For **many workers**, TPE may be slow to converge; consider `RandomSampler` in early phases or QMC for coverage.

### 10.7 Model Saving
```python
# Save the best model after optimization
best_trial = study.best_trial
best_params = best_trial.params
final_model = train_final(best_params)
joblib.dump(final_model, "best_model.pkl")
```

## 12. Common Pitfalls

| Pitfall | Fix |
| :--- | :--- |
| No `log=True` on lr/gamma | Always use `log=True` for exponential params |
| Too few trials (< 20) with TPE | Use `RandomSampler` or bump `n_startup_trials` |
| Not using pruning | Report intermediate values & raise `TrialPruned` |
| Study lost after crash | Use `storage=...` + `load_if_exists=True` |
| Changing param names mid-study | Keep parameter schemas stable |
| Tuning too many params with small budget | Reduce search space or increase trials |
| **Suggesting a param but not using it** | Remove unused suggests — pollutes importances |
| **Returning a non-scalar** | Return `float` (or tuple for multi-objective) |
| **Silent exceptions** | Set `optuna.logging.set_verbosity(optuna.logging.DEBUG)` |
| **SQLite with many workers** | Switch to PostgreSQL |
