```python
import numpy as np
import pandas as pd
```

Load Dataset

```python
df = pd.read_csv('student_clustering.csv')
df.head()
```

```python
df.shape
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['cgpa'], df['iq'])
plt.show()
```

```python
from sklearn.cluster import KMeans

wcss = []
for i in range(1,20):
  km = KMeans(n_clusters = i)
  km.fit_predict(df[['cgpa', 'iq']])
  wcss.append(km.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1, 20), wcss)
plt.show()
```

```python
X = df.iloc[:,:].values
km = KMeans(n_clusters=4)
y_means = km.fit_predict(X)
y_means
```

```python
plt.scatter(X[y_means == 0,0], X[y_means == 0,1])
plt.scatter(X[y_means == 1,0], X[y_means == 1,1])
plt.scatter(X[y_means == 2,0], X[y_means == 2,1])
plt.scatter(X[y_means == 3,0], X[y_means == 3,1])
plt.scatter(km.cluster_centers_[:,0], km.cluster_centers_[:,1], s=100, c='red')

plt.show()
```

For 3D Data

```python
from sklearn.datasets import make_blobs

centroids = [(-5,-5,5),(5,5,-5),(3.5,-2.5,4),(-2.5,2.5,-4)]
cluster_std = [1,1,1,1]

X,y = make_blobs(n_samples=200,cluster_std=cluster_std,centers=centroids,n_features=3,random_state=1)
```

```python
X
```

```python
y
```

```python
import plotly.express as px
fig = px.scatter_3d(x=X[:,0],y=X[:,1],z=X[:,2])
fig.show()
```

```python
wcss = []

for i in range(1, 20):
  km = KMeans(n_clusters=i)
  km.fit_predict(X)
  wcss.append(km.inertia_)
```

```python
plt.plot(range(1,20), wcss)
plt.show()
```

```python
km = KMeans(n_clusters=4)
yhat = km.fit_predict(X)
yhat
```

```python
df = pd.DataFrame()

df['col1'] = X[:,0]
df['col2'] = X[:,1]
df['col3'] = X[:,2]
df['label'] = yhat
```

```python
df.head()
```

```python
import plotly.graph_objects as go
fig = px.scatter_3d(df, x='col1', y='col2', z='col3', color='label')

# 2. Correctly add the cluster centers as a new trace
fig.add_trace(
    go.Scatter3d(
        x=km.cluster_centers_[:, 0],
        y=km.cluster_centers_[:, 1],
        z=km.cluster_centers_[:, 2],
        mode='markers',
        marker=dict(color='red', size=10, symbol='diamond'),
        name='Centroids'
    )
)

# 3. Display the interactive plot
fig.show()
```
