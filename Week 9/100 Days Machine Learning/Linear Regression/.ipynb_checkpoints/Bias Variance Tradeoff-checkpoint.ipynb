# Bias-Variance Trade-off

---
![Bias-Variance Trade-off](https://i.pinimg.com/1200x/68/12/38/681238b66b550b548f5f6863e20c7f1f.jpg)

---

The bias-variance tradeoff is <mark>the balancing act between a model being too simple (high bias) and too complex (high variance) to minimize total prediction error</mark>

### What is Bias
- Error caused by overly simple assumptions in the learning algorithm
- High bias leads to **underfitting**, meaning the model fails to capture important patterns in the data
- e.g., using a straight line (linear regression) to model curved or non-linear data

### What is Variance
- Error caused by the model being overly sensitive to small flactuations and noise in the training
- High variance leads to **overfitting**, meaning the model learns the training data too well$-$including random noise$-$and performs poorly on unseen data
- e.g., a deep decision tree or unconstrained neural tree that memorizes training points instead of learning general rules
---

#### The Tradeoff and Total Error

Total prediction error is broken down into three parts:
$$Total Error = Bias^2 + Variance + Irreducible \text{ Error} $$

- As you increase model complexity, **bias decreases** because the model can fit the data better
- At the same time, **variance increases** bevcause the model adapts too closely to specific training samples
- The goal is to find the optimal "sweet spot" of complexity where the sum of bias and variance is at its lowest

---

|Manage the Tradeoff|
|:-----------------:|
|**To reduce bias**, increase model complexity, and add more features, or use a more flexible algorithm|
|**To reduce variance**, add more training data, use **regularization**, or apply **ensemble** methods like *bagging*|

---

|Irreducible Error|
|:-----------------:|
|This part is due to the noisiness of the data itself. The only way to reduce this part of the error is to clean up the data (e.g., fix the data sources, such as broken sensors, or detect and remove outliers)|

---

|NOTE|
|:-----------------:|
|Increasing  a model's complexity will typically increases its variance and reduce its bias. Conversely, reducing a model's complexity increase its bias and reduces its variance. This is why it is called a trade-off|
|Generalization&rarr;a model's ability to perform accurately on new, nuseen data that was not used during training|
