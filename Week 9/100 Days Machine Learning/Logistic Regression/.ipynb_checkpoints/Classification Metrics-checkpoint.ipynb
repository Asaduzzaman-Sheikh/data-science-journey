# 1. Accuracy Score
$${accuracy \quad score} = \frac{\text{No. correct predictions}}{\text{Total number of predictions}}$$

## 1.1 How Much Accuracy Score is Good
There is no universal answer to this question. It depends on the problem we're solving

## 1.2 Problem With Accuracy Score
It does not tell about the *error type* $-$ *type I or type II*. It is also misleading when the class is imbalanced (dataset imbalance)

# 2. Confusion Matrix
A way to tell that how much your model making which type of errors
![Confusion Matrix](https://accessibleai.dev/img/Metrics/Classification/ConfusionAll.png)

We can calculate the accuracy score from the confusion matrix, but we cannot calculate the confusion matrix from the accuracy metric

$$Accuracy = \frac{TP + TN}{TP + FN + FP + TN}$$
>NOTE:
>
>**FALSE POSITIVE &rarr; TYPE I ERROR**
>
>**FALSE NEGATIVE &rarr; TYPE II ERROR**


# 3. Precision

What proportion of predicted positives is correctly classified
$$Precision = \frac{TP}{TP + FP}$$

So, when two models' accuracy is equal, and *precision* is the concern based on the problem &rarr; the one with the higher *precision* score is the better model.

# 4. Recall
What proportion of actual positives is correctly classified
$$Recall = \frac{TP}{TP + FN}$$

So, when two models' accuracy is equal, and *recall* is the concern based on the problem &rarr; the one with the higher *recall* score is the better model.


>NOTE
>
>When *type I error* is the concern &rarr; choose the *precision* with higher score
>
>When *type II error* is the concern &rarr; choose the *recall* with higher score

# 5. F1 Score
When it is difficult to determine which one is the main concern (*precision or recall*) &rarr; here comes the **F1 Score**; which is a *harmonic mean* of precision and recall, and it always penalizes the lower one (precision or recall) and stays close to that value
$$F1 = \frac{2\times P\times R}{P + R}$$
